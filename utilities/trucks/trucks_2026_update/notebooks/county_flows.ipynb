```python
# Heatmap diff OD flows
diff = ((mtc_county_od_17 - mtc_county_od_16))

plt.figure(figsize=(10, 8))
ax = sns.heatmap(
    diff,
    cmap="PuOr",
    center=0,
    annot=True,
    fmt=",.0f",
    cbar_kws={"label": "Trip Difference"}
)

# Move destination county labels to top
ax.xaxis.tick_top()
ax.xaxis.set_label_position('top')
ax.set_xticklabels(ax.get_xticklabels(), rotation=45, ha="left")

plt.title("County-to-County Trucks OD Flow Difference\n(TM-1.7 - TM-1.6)", pad=40)
plt.xlabel("Destination County")
plt.ylabel("Origin County")

plt.tight_layout()
plt.show()
```
